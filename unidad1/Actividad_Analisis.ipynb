## Celda 7 · Código de partida (del gitbook)

```python
import requests
import json
import pandas as pd

# Hacemos la solicitud a la API
url = "https://www.zaragoza.es/sede/servicio/restaurante.geojson?srsname=wgs84&rows=2000"
response = requests.get(url)

# Convertimos la respuesta a JSON
data = response.json()

# Mostramos las claves principales del JSON
print("claves")
print(data.keys())


# Paso 2: Extraer la lista de features (cada restaurante)
features = data["features"]
print(features)

# Paso 3: De cada restaurante, extraer solo el diccionario 'properties'
records = [f["properties"] for f in features] # esto equivale a las siguientes lineas ↓↓↓↓↓
# records = [] 
# for f in feature:
#   records.append(f["properties"])
df = pd.DataFrame(records)
print("Dimensiones del Data frame: ",df.shape)
```

## Celda 8 · Paso 0: preparar columnas
La API **no tiene** columna `tipo` y el teléfono viene dentro de un diccionario.

| Enunciado | Columna real |
|---|---|
| nombre | `title` |
| dirección | `streetAddress` |
| localidad | `addressLocality` |
| código postal | `postalCode` |
| teléfono | `tel` → creamos `telefono` |
| tipo | creamos `tipo` desde `tenedores` |

Donde el enunciado diga **"Bar"**, usa **"2 tenedores"**.

```python
# Tu código aquí
df["tipo"] = df["tenedores"].map({1: "restaurant", 2: "bar", 3: "cateferia"})
#Creo una nueva columna utilizando un valor de otra y asigno u "tipo" ↑↑↑↑↑↑
df["telefono"] = df["tel"].str.get("tel")


print(df[df["tipo"]=="bar"])
```
