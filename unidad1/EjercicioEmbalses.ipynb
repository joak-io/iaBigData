Trabajaréis con un CSV que recoge el estado actual de 444 embalses españoles. Los datoss proceden del boletín del MITECO y se publican en [mapasdeembalses.es](https://mapasdeembalses.es).

## Carga de datos

Ejecutad la siguiente celda para descargar los datos y cargarlos en un DataFrame de pandas.

```python
import requests
import pandas as pd
from io import StringIO

url = "https://mapasdeembalses.es/descargas/estado-actual.csv"
respuesta = requests.get(url)
respuesta.encoding = "utf-8"

# Cargar el CSV usando punto y coma como separador
df = pd.read_csv(StringIO(respuesta.text), sep=";")
df.head()
```

## 3. Modificación del DataFrame

8. Crea un nuevo DataFrame sin los embalses que no tienen datos de esta semana. ¿Por qué no sería adecuado rellenar esos datos ausentes con un 0?
9. Añade una columna que indique si el porcentaje de llenado de cada embalse está por debajo o por encima de la media nacional.
10. Ordena la tabla de mayor a menor porcentaje de llenado.

Escribe y ejecuta tu código en la celda siguiente. Para la justificación del ejercicio 8, puedes añadir una celda Markdown.

8- Veo, creo y considero que no sería optimo puesto que, el hecho de que no hayan datos no necesariamente significa que estén vacios y a la hora de analizar los datos el tenerlos en 0 seria bastante equivocado creo. 

```python
# 8. Crea un DF sin embalses desactualizados
print ("Ej 8: ")
df_actualizado =df.query("freshness_status == 'fresh'")
print(df_actualizado)

# 9. Compara cada porcentaje con la media nacional
print ("Ej 9: ")
mediaNacional = df_actualizado["fill_percentage"].mean()
print("Media Nacional de llenado: ", round(mediaNacional,2), "%")
df_actualizado.loc[df_actualizado["fill_percentage"] < mediaNacional, " vs media"] = "por debajo"

# 10. Ordena de mayor a menor porcentaje de llenado
print ("Ej 10: ")
df_mayorAmenor = df.sort_values(by='fill_percentage', ascending=False)
print(df_mayorAmenor[["id", "name","fill_percentage"]])
```

## 4. Análisis estadístico

11. Calcula el porcentaje medio de llenado a nivel nacional.
12. ¿Qué provincia tiene, de media, los embalses más llenos? ¿Y cuál tiene los más vacíos?

```python
# 11. Media nacional del porcentaje de llenado
print("Ej 11: ")
mediaNacional=df_actualizado["fill_percentage"].mean()
print("Media Nacional del procentaje de llenado: ", round(mediaNacional,2) , "%")

# 12. Provincias con la media más alta y más baja
print("Ej 12: ")
mediaProvincias=df.groupby("province")["fill_percentage"].mean()
provMayor= mediaProvincias.index[mediaProvincias.argmax()]
provMenor= mediaProvincias.index[mediaProvincias.argmin()]

print("Provincia con mayor porcentaje: ", provMayor)
print("Provincia con menor porcentaje: ", provMenor)

# 13. Número de embalses por cuenca hidrográfica
print("Ej 13: ")
cuencas=df.groupby("basin").size().sort_values(ascending=False)
print(cuencas)

# 14. Cinco embalses con mayor capacidad máxima
print("\nEj 14: ")
embalsesMaximos=df_actualizado.nlargest(5, "maximum_capacity_hm3")
print("Emablses con mayor capacidad máxima\n" , embalsesMaximos[["id","name","province","maximum_capacity_hm3"]])
```
